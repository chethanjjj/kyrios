## Session-level datasets

- [x] session.laps — Lap-by-lap data (already using)
    - Lap times, sector times, positions, tire info, pit stops, etc.
- [x] session.results — Final session results
    - Driver names, teams, finishing positions, grid positions, points, status (finished/DNF/etc.)
- [ ] session.drivers — Driver information
    - Driver numbers, abbreviations, full names, team associations
- [ ] session.event — Event metadata
    - Event name, location, country, date, round number, circuit info

## Time-series telemetry data (high-frequency)

- [ ] session.car_data — Car telemetry (via get_car_data())
    - Speed, RPM, throttle, brake, gear, DRS status, normalized track position
    - Sampled at ~10 Hz (multiple points per second)
- [ ] session.pos_data — Position data (via get_pos_data())
    - X, Y coordinates on track, status (on track/in pit), normalized position
    - Also high-frequency time-series
- [ ] session.weather_data — Weather conditions
    - Air temperature, track temperature, humidity, pressure, wind speed/direction
    - Time-series during the session

## Session status and control

- [ ] session.track_status — Track status changes
    - Flag status (green/yellow/red), safety car deployments, virtual safety car periods
    - Timestamps for status changes
- [ ] session.session_status — Session lifecycle
    - Session start/end times, status (not started/running/finished)
- [ ] session.race_control_messages — Race control communications
    - Penalties, investigations, warnings, restart announcements
    - Timestamped messages
  
## Additional metadata

- [ ] session.event.schedule — Session schedule
    - Start times for practice, qualifying, race sessions
- [ ] session.event.weather — Weather forecast (if available)
  - Pre-session weather predictions

## Quick Reference Summary

### Direct Attributes (DataFrames/Series):
- `session.laps` - Lap-by-lap data (DataFrame)
- `session.results` - Final session results (DataFrame)
- `session.drivers` - List of driver numbers
- `session.event` - Event metadata (Series)
- `session.weather_data` - Weather conditions (DataFrame)
- `session.track_status` - Track status changes (DataFrame)
- `session.session_status` - Session lifecycle (DataFrame)
- `session.race_control_messages` - Race control messages (DataFrame)
- `session.name` - Session name (string)
- `session.date` - Session date (datetime)

### Methods (require parameters):
- `session.get_car_data(driver)` - Car telemetry for a driver (or list of drivers)
- `session.get_pos_data(driver)` - Position data for a driver (or list of drivers)

### Useful Laps Methods:
- `session.laps.pick_driver('44')` - Filter laps for specific driver
- `session.laps.pick_fastest()` - Get fastest lap
- `session.laps.slice_by_time(start, end)` - Get laps in time window
- `session.laps.pick_team('Mercedes')` - Filter laps by team
